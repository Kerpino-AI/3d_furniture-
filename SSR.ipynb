تغییر سایز تصاویر به [484, 648]

```python
import os
from PIL import Image

folder_path = "/content/kerpoo"
output_folder = "/content/drive/MyDrive/ssr/SSR-code/data/custom_images"

if not os.path.exists(output_folder):
    os.makedirs(output_folder)

valid_extensions = [".jpg",".jpeg",".png",".bmp",".tiff"]

images = [img for img in os.listdir(folder_path)
          if os.path.splitext(img)[1].lower() in valid_extensions]
images.sort()

counter = 1

for img_name in images:
  img_path = os.path.join(folder_path, img_name)
  try:
      with Image.open(img_path) as img:
        new_img = img.resize((648, 484), Image.Resampling.LANCZOS)

        new_name = f"kerpoo-SSR-{counter:04d}.jpg"
        output_path = os.path.join(output_folder, new_name)
        new_img.save(output_path,"JPEG")
        print(f"{new_name} saved.")

        counter += 1
  except Exception as e:
    print(e)

print("converting images is finished!")

```
