# **For Croping Images to distinct identity and using COLMAP and Structure-from-Motion (SfM) and Multi-View Stereo (MVS) pipeline**

```python
import os
import torch
import requests
from PIL import Image
from diffusers import DiffusionPipeline, EulerAncestralDiscreteScheduler

# لطفا این آدرس ها رو متناسب با سیستمت تغییر بذه
os.makedirs("/content/drive/MyDrive/zero123_extended/colmap_project", exist_ok=True)
OUTPUT_DIR = "/content/drive/MyDrive/zero123_extended/colmap_project"
IMAGES_DIR = os.path.join(OUTPUT_DIR, "images")
os.makedirs(IMAGES_DIR, exist_ok=True)


pipeline = DiffusionPipeline.from_pretrained("sudo-ai/zero123plus-v1.2",
                                             custom_pipeline = "sudo-ai/zero123plus-pipeline",
                                             torch_dtype = torch.float16)

pipeline.scheduler = EulerAncestralDiscreteScheduler.from_config(pipeline.scheduler.config, timestep_spacing = "trailing")

pipeline.to("cuda:0")
# pipeline.to("cpu")


# لطفا آدرس تصویر را اینجا وارد کنید
img_url = "/content/drive/MyDrive/zero123_extended/image_0001.jpg"

cond = Image.open(img_url)


# شایان جان برای دقت بالاتر تا 90 میتونی inference رو تنطیم کنی
tiled_img = pipeline(cond, num_inference_step = 75).images[0]
tiled_img_path = os.path.join(IMAGES_DIR, "tiled_output.png")
tiled_img.save(tiled_img_path)
print(f"Saved tiled image to {tiled_img_path}")

cols, rows = 2, 3   # change this in other situation
tile_width, tile_height = tiled_img.size[0]//cols, tiled_img.size[1]//rows


for idx, (row, col) in enumerate([r , c] for r in range(rows) for c in range(cols)):
    left, upper = col * tile_width, row * tile_height
    right, lower = left + tile_width, upper + tile_height
    tile = tiled_img.crop((left, upper, right, lower))


    tile_path = os.path.join(IMAGES_DIR, f"view_{idx:02d}.png")
    tile.save(tile_path)
    print(f"Saved tile {idx} to {tile_path}")

```

# **Neural Radiance Field For 3D Reconstruction**

Creating transformer.json

Azimuth (relative to input view): 30, 90, 150, 210, 270, 330.
v1.1 Elevation (absolute): 30, -20, 30, -20, 30, -20.
v1.2 Elevation (absolute): 20, -10, 20, -10, 20, -10.
v1.2 Field of View (absolute): 30°.

```python
import json
import math
import numpy as np

def create_transform(azimuth_deg, elevation_deg, radius = 1.0):

    # to radian

    az = math.radians(azimuth_deg)
    el = math.radians(elevation_deg)

    # creating camera position in spherical coordinates

    x = radius * math.cos(el) * math.sin(az)
    y = radius * math.sin(el)
    z = radius * math.cos(el) * math.cos(az)

    transform = np.eye(4).tolist()
    transform[0][3] = x
    transform[1][3] = y
    transform[2][3] = z

    return transform


angles = [30, 90, 150, 210, 270, 330 ]
elevations = [20, -10, 20, -10, 20, -10]

frames = []

for i, (az, el) in enumerate(zip(angles, elevations)):
    transform = create_transform(az, el, radius = 1.0)
    frames.append({"file_path":f"/content/drive/MyDrive/zero123_extended/nerf/images/view_{i:02d}.png",
                   "transform_matrix": transform})

data = {
    "camera_angle_x" : 0.5235987755982988,
    "frames" : frames,
}

with open("/content/drive/MyDrive/zero123_extended/nerf/transforms.json", "w") as f :
    json.dump(data, f, indent = 4)

print("trnsform.json is created in nerf directory")
```

```python
import cv2
import torch
import numpy as np
import os
import open3d as o3d
from midas.midas_net import MidasNet
from midas.transforms import Resize, NormalizeImage, PrepareForNet
import torchvision.transforms as transforms

# بارگذاری مدل MiDaS
model_path = "model-f6b98070.pt"
model = MidasNet(model_path, non_negative=True)
model.eval()
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model.to(device)

# تعریف تبدیل‌ها (transforms) برای MiDaS
transform = transforms.Compose([
    Resize(384, 384),  # تغییر اندازه تصاویر به 384x384 (می‌توانید بر حسب نیاز تغییر دهید)
    NormalizeImage(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
    PrepareForNet()
])

def process_image(image_path):
    # بارگذاری تصویر
    img = cv2.imread(image_path)
    if img is None:
        print("Error reading:", image_path)
        return None, None
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    input_tensor = transform(img_rgb).unsqueeze(0).to(device)
    with torch.no_grad():
        depth = model(input_tensor)
    depth_np = depth.squeeze().cpu().numpy()
    return depth_np, img_rgb

# مسیر فولدر تصاویر (فرض می‌کنیم تصاویر در فولدر "images" قرار دارند)
image_folder = "images"
image_files = [os.path.join(image_folder, f) for f in os.listdir(image_folder) if f.lower().endswith(('.png','.jpg','.jpeg'))]

print("Found images:", image_files)

# پردازش تصاویر و ذخیره نقشه‌های عمق
depth_maps = []
for path in image_files:
    d, _ = process_image(path)
    if d is not None:
        depth_maps.append(d)
print("Extracted depth maps for", len(depth_maps), "images.")
```

```python
import torch
import cv2
import numpy as np
import open3d as o3d
import os

# انتخاب نوع مدل: "DPT_Large" یا "MiDaS_small" یا "DPT_Hybrid"
model_type = "DPT_Large"  # یا مدل دلخواهتان

# بارگذاری مدل از torch.hub
midas = torch.hub.load("intel-isl/MiDaS", model_type)
midas.eval()
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
midas.to(device)

# بارگذاری تبدیل‌های مورد نیاز
midas_transforms = torch.hub.load("intel-isl/MiDaS", "transforms")
transform = midas_transforms.small_transform if model_type == "MiDaS_small" else midas_transforms.dpt_transform

# تابع پردازش تصویر
def process_image(image_path):
    img = cv2.imread(image_path)
    if img is None:
        print("Error reading:", image_path)
        return None, None

    # تبدیل BGR به RGB و نرمال‌سازی تصویر
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB).astype(np.float32) / 255.0

    # اعمال transform و اطمینان از ابعاد صحیح
    input_tensor = transform(img_rgb).unsqueeze(0).to(device)

    # بررسی ابعاد قبل از فرستادن به مدل
    if input_tensor.dim() == 5:  # اگر ابعاد اضافی دارد، آن را حذف کن
        input_tensor = input_tensor.squeeze(1)

    with torch.no_grad():
        depth = midas(input_tensor)

    depth_np = depth.squeeze().cpu().numpy()
    return depth_np, img_rgb

# استفاده از تصاویر آپلود شده در فولدر "images"
image_folder = "/content/images"
image_files = [os.path.join(image_folder, f) for f in os.listdir(image_folder) if f.lower().endswith(('.png', '.jpg', '.jpeg'))]

print("Found images:", image_files)

depth_maps = []
for path in image_files:
    d, _ = process_image(path)
    if d is not None:
        depth_maps.append(d)

print("Extracted depth maps for", len(depth_maps), "images.")
```

```python
import cv2
import numpy as np
import open3d as o3d
import os

# مسیرها برای ذخیره خروجی
depth_folder = "/content/depth_maps"
pcd_folder = "/content/point_clouds"
os.makedirs(depth_folder, exist_ok=True)
os.makedirs(pcd_folder, exist_ok=True)

# ذخیره نقشه‌های عمق
def save_depth_maps(depth_maps):
    for i, depth in enumerate(depth_maps):
        depth_path = os.path.join(depth_folder, f"depth_{i:03d}.png")
        normalized_depth = (depth - depth.min()) / (depth.max() - depth.min())  # نرمال‌سازی
        cv2.imwrite(depth_path, (normalized_depth * 255).astype(np.uint8))
        print(f"Depth map saved: {depth_path}")

# تبدیل نقشه عمق به ابر نقاط
def depth_to_point_cloud(depth, img, scale=1.0):
    h, w = depth.shape
    fx = fy = max(h, w) * scale  # فاصله کانونی تقریبی
    cx, cy = w / 2, h / 2

    points = []
    colors = []
    for y in range(h):
        for x in range(w):
            d = depth[y, x]
            if d > 0:  # حذف نقاط با عمق صفر
                z = d
                x3d = (x - cx) * z / fx
                y3d = (y - cy) * z / fy
                points.append((x3d, y3d, z))
                colors.append(img[y, x] / 255.0)

    pcd = o3d.geometry.PointCloud()
    pcd.points = o3d.utility.Vector3dVector(np.array(points))
    pcd.colors = o3d.utility.Vector3dVector(np.array(colors))
    return pcd

# ذخیره ابر نقاط
def save_point_clouds(depth_maps, image_files):
    for i, (depth, img_path) in enumerate(zip(depth_maps, image_files)):
        img = cv2.imread(img_path)
        img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

        pcd = depth_to_point_cloud(depth, img_rgb)
        pcd_path = os.path.join(pcd_folder, f"point_cloud_{i:03d}.ply")
        o3d.io.write_point_cloud(pcd_path, pcd)
        print(f"Point cloud saved: {pcd_path}")

# ذخیره خروجی‌ها
save_depth_maps(depth_maps)
save_point_clouds(depth_maps, image_files)

print("All depth maps and point clouds are saved successfully!")
```

```python
import cv2
import numpy as np
import open3d as o3d
import os

# مسیرهای ذخیره خروجی
depth_folder = "/content/depth_maps"
pcd_folder = "/content/point_clouds"
os.makedirs(depth_folder, exist_ok=True)
os.makedirs(pcd_folder, exist_ok=True)

# ذخیره نقشه‌های عمق
def save_depth_maps(depth_maps):
    for i, depth in enumerate(depth_maps):
        depth_path = os.path.join(depth_folder, f"depth_{i:03d}.png")
        normalized_depth = (depth - depth.min()) / (depth.max() - depth.min())  # نرمال‌سازی
        cv2.imwrite(depth_path, (normalized_depth * 255).astype(np.uint8))
        print(f"Depth map saved: {depth_path}")

# تبدیل نقشه عمق به ابر نقاط
def depth_to_point_cloud(depth, img, scale=1.0):
    h, w = depth.shape
    img = cv2.resize(img, (w, h))  # همگام‌سازی ابعاد تصویر و نقشه عمق

    fx = fy = max(h, w) * scale  # فاصله کانونی تقریبی
    cx, cy = w / 2, h / 2

    points = []
    colors = []
    for y in range(h):
        for x in range(w):
            d = depth[y, x]
            if d > 0:  # حذف نقاط با عمق صفر
                z = d
                x3d = (x - cx) * z / fx
                y3d = (y - cy) * z / fy
                points.append((x3d, y3d, z))
                colors.append(img[y, x] / 255.0)

    pcd = o3d.geometry.PointCloud()
    pcd.points = o3d.utility.Vector3dVector(np.array(points))
    pcd.colors = o3d.utility.Vector3dVector(np.array(colors))
    return pcd

# ذخیره ابر نقاط
def save_point_clouds(depth_maps, image_files):
    for i, (depth, img_path) in enumerate(zip(depth_maps, image_files)):
        img = cv2.imread(img_path)
        img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

        pcd = depth_to_point_cloud(depth, img_rgb)
        pcd_path = os.path.join(pcd_folder, f"point_cloud_{i:03d}.ply")
        o3d.io.write_point_cloud(pcd_path, pcd)
        print(f"✅ Point cloud saved: {pcd_path}")

# ذخیره خروجی‌ها
save_depth_maps(depth_maps)
save_point_clouds(depth_maps, image_files)

print("All depth maps and point clouds are saved successfully!")
```

```python
import open3d as o3d
import numpy as np
import os

# مسیر ابر نقاط ورودی
pcd_folder = "/content/point_clouds"
output_pcd_path = "/content/registered_point_cloud.ply"

# خواندن تمام فایل‌های ابر نقاط
pcd_files = sorted([os.path.join(pcd_folder, f) for f in os.listdir(pcd_folder) if f.endswith(".ply")])
print(f"Found {len(pcd_files)} point clouds for registration.")

# بررسی تعداد فایل‌ها
if len(pcd_files) < 2:
    print("At least two point clouds are needed for registration!")
    exit()

# بارگذاری اولین ابر نقاط به عنوان مرجع
pcd_ref = o3d.io.read_point_cloud(pcd_files[0])
pcd_combined = pcd_ref

# تابع رجیستر کردن دو ابر نقاط با روش ICP
def register_point_clouds(source, target, threshold=0.02):
    icp_result = o3d.pipelines.registration.registration_icp(
        source, target, threshold,
        np.identity(4),
        o3d.pipelines.registration.TransformationEstimationPointToPoint(),
        o3d.pipelines.registration.ICPConvergenceCriteria(max_iteration=200)
    )
    return icp_result.transformation

# رجیستر کردن تمام ابر نقاط به اولین ابر نقاط
for i in range(1, len(pcd_files)):
    pcd_src = o3d.io.read_point_cloud(pcd_files[i])

    print(f"Registering {pcd_files[i]} to reference point cloud...")

    transformation = register_point_clouds(pcd_src, pcd_combined)
    pcd_src.transform(transformation)

    # ترکیب ابر نقاط رجیستر شده
    pcd_combined += pcd_src

# فیلتر کردن نویزها و کاهش تراکم ابر نقاط
pcd_combined = pcd_combined.voxel_down_sample(voxel_size=0.005)

# ذخیره‌ی ابر نقاط ترکیبی در فایل نهایی
o3d.io.write_point_cloud(output_pcd_path, pcd_combined)

print(f"Registered point cloud saved: {output_pcd_path}")
```

```python
import open3d as o3d

# بارگذاری ابر نقاط ثبت‌شده
input_pcd_path = "/content/registered_point_cloud.ply"
output_mesh_path = "/content/reconstructed_mesh.ply"

# خواندن ابر نقاط
pcd = o3d.io.read_point_cloud(input_pcd_path)

# محاسبه نرمال‌ها (برای بازسازی مش لازم است)
pcd.estimate_normals(search_param=o3d.geometry.KDTreeSearchParamHybrid(radius=0.1, max_nn=30))

# بازسازی مش با روش Poisson
mesh, densities = o3d.geometry.TriangleMesh.create_from_point_cloud_poisson(pcd, depth=8)

# فیلتر کردن نویزهای مدل
vertices_to_remove = densities < np.quantile(densities, 0.05)  # حذف 5% از نویزها
mesh.remove_vertices_by_mask(vertices_to_remove)

# ذخیره مدل سه‌بعدی مش‌بندی شده
o3d.io.write_triangle_mesh(output_mesh_path, mesh)

print(f"Mesh saved: {output_mesh_path}")
```
